```python
# ⚠️ __init__ не создаёт объект (это делает __new__), а лишь настраивает его начальное состояние.
# __init__(self, ...) -  Инициализация после создания объекта
# __getitem__(self, key) Чтение по индексу/ключу
# __setitem__(self, key, val) Запись по индексу/ключу
```

```python
# Эмулирует поведение списка с собственной внутренней логикой.

class SafeList:
    def __init__(self, items):
        self._data = list(items)
    def __getitem__(self, idx):
        return self._data[idx]
    def __setitem__(self, idx, val):
        self._data[idx] = val

```

```python
lst = SafeList([10, 20, 30])
lst[1] = 99
print(lst[1])  # 99
print(lst)  # 99
```

```python
lst = SafeList((10, 20, 30))
lst[1] = 99
print(lst[1])  # 99
```

```python
lst = SafeList('asavgxcds')
lst[1] = 99
print(lst[1])  # 99
```

```python
# 2️⃣ Словарь с дефолтным значением
# Аналог collections.defaultdict, но с явной реализацией.
class DefaultDict:
    def __init__(self, default=None):
        self._store = {}
        self._default = default
    def __getitem__(self, key):
        return self._store.get(key, self._default)
    def __setitem__(self, key, val):
        self._store[key] = val
```

```python
d = DefaultDict("N/A")
d["name"] = "Alice"
print(d["name"], d["age"])  # Alice N/A
```

```python
# 3️⃣ Матрица (2D-индексация через кортеж)
# Позволяет использовать синтаксис mat[i, j].

class Matrix:
    def __init__(self, rows, cols, fill=0):
        self.rows, self.cols = rows, cols
        self._data = [[fill]*cols for _ in range(rows)]
    def __getitem__(self, pos):
        i, j = pos
        return self._data[i][j]
    def __setitem__(self, pos, val):
        i, j = pos
        self._data[i][j] = val
```

```python
m = Matrix(3, 3)
m[1, 2] = 42
print(m[1, 2])  # 42
```

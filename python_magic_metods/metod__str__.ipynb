```python
# Метод __str__(self) отвечает за человекочитаемое строковое представление объекта.
# Он автоматически вызывается в трёх случаях:
# str(obj)
# print(obj)
# f-строки и format(): f"{obj}", "{}".format(obj)
```

```python
# 🟢 1. Базовый пример: str() и print()

class User:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def __str__(self):
        # ⚠️ Обязательно возвращаем str, а не печатаем!
        return f"👤 {self.name} ({self.role})"
```

```python
u = User("Алексей", "Admin")

print(str(u))  # 👤 Алексей (Admin)
print(u)       # 👤 Алексей (Admin)  ← print() внутри вызывает str(u)
```

```python
# 🟢 2. f-строки и .format() (тоже вызывают __str__)
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def __str__(self):
        return f"{self.name} — {self.price:.2f} ₽"
```

```python
p = Product("Ноутбук", 85000.5)

print(f"Товар: {p}")           # Товар: Ноутбук — 85000.50 ₽
print("Выбрано: {}".format(p)) # Выбрано: Ноутбук — 85000.50 ₽
```

```python
# humps пакет помогает конвертировать строки (и ключи словарей) между разными 
# стилями написания: snake_case, camelCase, pascalCase.
```

```python
import humps
```

```python
import humps

print(humps.camelize("jack_in_the_box"))  # jackInTheBox
print(humps.decamelize("rubyTuesdays"))  # ruby_tuesdays
print(humps.pascalize("red_robin"))  # RedRobin
print(humps.kebabize("white_castle"))  # white-castle
```

```python
# Converting dictionary keys
import humps

array = [{"attrOne": "foo"}, {"attrOne": "bar"}]
print(humps.decamelize(array)) # [{"attr_one": "foo"}, {"attr_one": "bar"}]

array = [{"attr_one": "foo"}, {"attr_one": "bar"}]
print(humps.camelize(array))  # [{"attrOne": "foo"}, {"attrOne": "bar"}]

array = [{'attr_one': 'foo'}, {'attr_one': 'bar'}]
print(humps.kebabize(array))  # [{'attr-one': 'foo'}, {'attr-one': 'bar'}]

array = [{"attr_one": "foo"}, {"attr_one": "bar"}]
print(humps.pascalize(array))  # [{"AttrOne": "foo"}, {"AttrOne": "bar"}]
```

# Work with classes

```python
class A:
    def __init__(self, aa: str) -> None:
        self.aa = aa
        print(f"class A {self.aa}")


class B(A):
    def __init__(self, aa: str, bb: str) -> None:
        self.bb = bb
        super().__init__(aa)
        print(f"class B {aa}")
        print(f"class B bb {self.bb}")


class C(B):
    def __init__(self, aa: str, bb:str, cc:str) -> None:
        self.cc = cc
        super().__init__(aa, bb)
        print(f"class C {aa}")
        print(f"class C bb {bb}")
        print(f"class C cc {self.cc}")
        
```

```python
dd = C("Hello!", "Yes!", "No")
```

```python
dd = B("Hello!", "Yes!")
```

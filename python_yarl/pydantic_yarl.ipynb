```python
# yarl — библиотека для работы с URL 
# yarl (Yet Another URL Library) — библиотека от команды aio-libs (авторы aiohttp).
# Главная идея: URL — это иммутабельный объект с удобными методами для чтения и модификации,
# с автоматическим экранированием (percent-encoding). Именно yarl.URL используется внутри aiohttp.
```

```python
from yarl import URL
```

```python
# Из строки
url = URL("https://example.com/api/users?page=2&limit=10")
```

```python
# «С нуля» — экранирование произойдёт автоматически
url = URL.build(
    scheme="https",
    host="example.com",
    path="/api/users",
    query={"page": 2, "limit": 10},
)
```

```python
print(url)
```

```python
url = URL("https://user:pass@example.com:8080/api/users?page=2&limit=10#top")

url.scheme         # 'https'
url.host           # 'example.com'
url.port           # 8080
url.user           # 'user'
url.password       # 'pass'
url.path           # '/api/users'
url.query_string   # 'page=2&limit=10'
url.fragment       # 'top'

str(url)           # полная строка URL
url.human_repr()   # читаемая версия (с декодированными %-последовательностями)
```

```python
# Модификация — всегда новый объект
# URL иммутабелен: все методы with_* и update_query
# 
# возвращают новый URL, оригинал не меняется. Методы можно чейнить:
```

```python
url = URL("https://example.com/api")

url1 = url.with_scheme("http")              # смена схемы
print('1', url1)
url1 = url.with_host("api.example.com")     # смена хоста
print('2', url1)
url1 = url.with_port(8443)                  # смена порта
print('3', url1)
url.with_path("/v2/users")           # замена пути целиком
url.with_query({"page": 2})          # ПОЛНАЯ замена query
url.with_query(page=2)               # то же, kwargs
url2 = url.update_query(page=3)             # объединение с существующими
print('4', url2)
url3 = url2.with_port(8443)
print('5', url3)
url.with_fragment("section")         # якорь
url.with_user("admin").with_password("secret")
url.origin()                         # https://example.com — без пути/query/фрагмента
# url.with_query(None)                 # очистить query
print('2', url)
url.update_query                 # очистить query
print('3', url)
```
